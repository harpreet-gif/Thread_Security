## **Linear Regression**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
house_sizes = np.array([500, 750, 1000, 1200, 1500, 1700, 2000, 2250]).reshape(-1, 1)
house_prices = np.array([150, 175, 250, 300, 300, 400, 400, 500])
```

```python
X_train, X_test, y_train, y_test = train_test_split(house_sizes, house_prices, test_size = 0.20, random_state = 42)
```

```python
X_train
```

```python
X_test
```

```python
y_train, y_test
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
```

```python
print(f"MSE: {mse: .2f}")
print(f"R2 Score: {r2: .2f}")
```

```python
plt.scatter(house_sizes, house_prices, color = "Blue", label = 'actual data')
plt.plot(house_sizes, model.predict(house_sizes), color = 'red', label = 'Regression line')
plt.title("House price prediction")
plt.xlabel("House size (sq ft.)")
plt.ylabel("Price")
plt.legend()
plt.show()
```
