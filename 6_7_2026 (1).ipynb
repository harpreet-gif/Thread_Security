```python
# Function with parameter

def marks(maths, science):
  print("Marks in Maths:", maths)
  print("Marks in Science:", science)
marks(89, 85)
```

```python
# Default parameter

def hello(name = "User"):
  print("Welcome", name)

hello()
hello("Harpreet")
```

```python
# Positional parameter

def students(name1, name2):
  print("First student: ", name1)
  print("Second student: ", name2)

students("Preet", "Riya")
students("Riya", "Preet")
```

```python
# Keyword parameter

def marks(maths, english):
  print("Maths:", maths)
  print("English:", english)

marks(english = 91, maths = 89)
marks(maths = 89, english = 95)
```

```python
def sum(a, b, c):
  return a+b+c

result = sum(2, 6, 9)
print("The sum of 3 numbers is: ", result)
```

```python
def add(a, b):
  return a + b
add(5, 5)
```

```python
def add(a, b):
  return a + b

result = add(5, 5)
print("the sum is:", result)
```

```python
# *args parameter

def numbers(*args):
  print(args)
numbers(10, 20, 30, 40)
```

```python
def subjects(*args):
  print("Subjects: ",args)
subjects("Python", "Java", "AI", "Maths")
```

```python
def subjects(*args):
  for i in args:
    print(i)
subjects("Python", "Java", "AI", "Maths")
```

```python
# **kwargs parameter

def student(**kwargs):
  print(kwargs)
student(name='preet', age=19, course='CSE')
```

```python
def student(**kwargs):
  for key ,values in kwargs.items():
    print(key, ':' ,values)

student(name='preet', age=19, course='CSE')
```
