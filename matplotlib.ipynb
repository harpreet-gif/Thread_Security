```python
import matplotlib.pyplot as plt
```

## **1. Line Graph**

```python
x = [12, 45, 58, 65, 70, 84]
y = [100, 350, 450, 500, 550, 600]

plt.plot(x, y, color = 'cyan', marker = 'o', linestyle = '--')
plt.title("Simple line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## **Multi-line graph**

```python
x = [12, 45, 58, 65, 70, 84]
y1 = [100, 350, 450, 500, 550, 600]
y2 = [111, 146, 269, 389, 450, 555]

plt.plot(x, y1, color = 'blue', marker = 'o', linestyle = '-', label = 'line 1')
plt.plot(x, y2, color = 'orange', marker = 'o', linestyle = '-', label = 'line 2')
plt.title("Multi-line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

## **2. Bar chart**

```python
classes = ['A', 'B', 'C', 'D']
students = [34, 56, 45, 67]

plt.bar(classes, students, color = 'green', edgecolor = 'yellow', alpha = 0.7)
plt.title("Bar Chart")
plt.xlabel("Classes")
plt.ylabel("Number of Students")
plt.show()
```

## **3. Histogram**

```python
marks = [45, 50, 55, 60, 65, 70, 72, 75, 80, 82, 85, 90, 95]
plt.hist(marks, bins=5, color='blue', edgecolor='black')
plt.title("Histogram of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

## **4. Scatter plot**

```python
hours = [1, 2, 3, 4, 5]
marks = [35, 45, 55, 70, 85]

plt.scatter(hours, marks,color="m",
    s=100,
    marker="*",
    alpha=1.0)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

## **5. Pie chart**

```python

subjects = ["Math", "Science", "English", "Computer"]
marks = [30, 25, 20, 25]

explode = [0., 0.1, 0.1, 0.1]
colors = ['yellow', 'm', 'blue', 'red']

plt.pie(marks, labels=subjects, autopct="%1.1f%%", explode = explode, colors = colors)

plt.title("Marks Distribution by Subject")

plt.show()
```
