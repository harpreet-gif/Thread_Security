```python
# create a function that multiply each element of the list

def elements(*args):
  product = 1

  for i in args:
    product = product * i
    print(product)

elements([2, 3, 4, 5, 1])
print()
print("Product of each element in the list: ")
elements(2, 3, 4, 5, 1)
```

```python
# Reverse string

def reverse():
  string = input("Enter a string: ")
  reverse_string = string[::-1]
  print("Reverse of string is:",reverse_string)

reverse()
```

```python
def string():
  sentence = input("Enter a sentence: ")

  a = e = i = o = u = 0
  non_vowel = 0

  for ch in sentence:
    if ch == "a":
     a = a+1
    elif ch == "e":
      e += 1
    elif ch == "i":
      i += 1
    elif ch == "o":
      o += 1
    elif ch == "u":
      u += 1
    elif ch.isalpha():
      non_vowel += 1

  print("a :", a)
  print("e :", e)
  print("i :", i)
  print("o :", o)
  print("u :", u)
  print("Non Vowel letters: ", non_vowel)

string()
```
