# **1. ATM Machine**

```python
balance = 2500

while True:
  print("1. Check Balance")
  print("2. Deposit")
  print("3. Withdraw")
  print("4. Exit")

  choice = input("Enter your choice: ")

  if choice == "check balance" or choice == "1":
    print("Current balance: ", balance)

  elif choice == "deposit" or choice == "2":
    amount = int(input("Enter your amount to deposit : " ))

    if amount > 0:
      balance += amount
      print(print("Balance after deposit amount: ", balance))
    else:
      print("Enter Positive amount")


  elif choice == "withdraw" or choice == "3":
    amount = int(input("Enter your amount to withdraw: "))

    if amount <= balance:
      balance -= amount
      print("Balance after withdraw amount:", balance)
    else:
      print("Withdrawal cannot exceed balance.")


  elif choice == "exit" or choice == "4":
    print("Exited.........")
    break

  else:
    print("Invalid choice!")
```

# **2. Password Validation**

```python
while True:
  password = input("Enter your password: ")

  has_upper = False
  has_lower = False
  has_digit = False

  if len(password) >= 8:
    for ch in password:
      if ch.isupper():
        has_upper = True
      if ch.islower():
        has_lower = True
      if ch.isdigit():
        has_digit = True

    print("Password Accepted")
    break

  else:
    print("Invalid password!")
    print("Password should contain:")
    print("At least 8 characters")
    print("one uppercase letter")
    print("one lowercase letter")
    print("one digit")



```

# **8. Find Duplicate Numbers**

```python
num = int(input("Enter numbers: "))

for i in range(1, 11):
  for j in range(i, 11):
    if num[i] == num[j]:
      print("Duplicate found")

```
