```python
income = int(input("Enter your Income:"))

if income >= 50000:
  credit_score = int(input("Enter credit score:"))

  if credit_score >= 750:
    print("Loan Approved")
  else:
    print("Loan rejected due to low credit score")

elif income <= 50000:
  print("Loan rejected due to low income")


```

```python
percentage = int(input("Enter percentage:"))

if percentage >= 65:

  aptitudeTest = input("Have you cleared the Aptitude Test? (Yes or NO)")

  if aptitudeTest == "yes":
    print("Selected for Technical Interview.")

  else:
    print("Rejected in Aptitude Round.")

elif percentage < 65:
  print("Not Eligible for Recruitment.")

```
