```python
num = 5
for i in range(5, 0, -1):
  for j in range(i):
    print(i, end = " ")
    num += 1

  print()
```

```python
num = 10

for i in range(1, 5):
  for j in range(i, 5):
    print(num, end = " ")
    num -= 1

  print()
```
