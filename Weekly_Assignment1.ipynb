### **5. Shopkeeper Bill**

```python
# prices
notebook_price = 35
pen_price = 12

# quantities
notebooks = 4
pens = 6

# (a) total cost of notebooks
total_notebook_cost = notebook_price * notebooks
print("Total cost of notebooks : ", total_notebook_cost)

# (b) total cost of pens
total_pen_cost = pen_price * pens
print("Total cost of pens : ", total_pen_cost)

# (c) grand total the customer has to pay.
total_cost = total_notebook_cost + total_pen_cost
print("Grand total to be paid : ", total_cost)

# (d) change returned if customer pays 250/-
payment = 250
change = payment - total_cost
print("Change to be returned : ", change)
```

### **6. Chocolate Distribution**

```python
chocolates = 57
students = 8

# (a) How many chocolates does each student get?
chocolates_per_student = chocolates // students
print("Chocolates per student : ", chocolates_per_student)

# (b) How many chocolates are left over after the distribution?
chocolates_left = chocolates % students
print("Chocolates left over : ", chocolates_left)

# (c) Donation box
money = 500
pen_price = 18

pens_bought = money // pen_price
print("Pens bought : ", pens_bought)

money_left = money % pen_price
print("Money left : ", money_left)
```

### **8. Weather Outfit Suggestion**

```python
# store temperature
temp = 45

if temp < 10:
  print("Wear a heavy coat and gloves")

elif temp <= 19:
  print("A light jacket should be fine.")

elif temp <= 30:
  print("T-shirt weather! Enjoy the day.")

else:
  print("Stay hydrated and wear light clothes.")

# extra warning
if temp > 40:
  print("Extreme heat! Avoid going out.")
```

### **9. Mini Student Report Card Generator**

```python
# (a) Student Details
student_name = "Riya"
student_class = "10th"
roll_number = 42
marks = [87, 79, 90, 75, 91]

# (b) calculate total and Percentage
total = marks[0] + marks[1] + marks[2] + marks[3] + marks[4]
percentage = (total/500)*100

# (c) print report card
print("=================================================")
print("                STUDENT REPORT CARD              ")
print("=================================================")
print("Name:", student_name)
print("Class:", student_class)
print("Roll Number:", roll_number)

print(" ")

print("--------------------------------------------------")
print("Maths              :", marks[0])
print("Science            :", marks[1])
print("English            :", marks[2])
print("History            :", marks[3])
print("Computer Science   :", marks[4])
print("--------------------------------------------------")

print(" ")

print("Total Marks:", total, "/500")
print("Percentage:", percentage, "%")

# (d) Grade Calculation
if percentage >= 90:
  print("Grade: A+")

elif percentage >= 80:
  print("Grade: A")

elif percentage >= 70:
  print("Grade: B")

elif percentage >= 60:
  print("Grade: C")

else:
  print("Grade: D")

# (e) Result
if marks[0] < 33 or marks[1] < 33 or marks[2] < 33 or marks[3] < 33 or marks[4] < 33:
  print("Result: FAIL - Improvement needed in one or more subjects.")

else:
  print("Result: PASS - Congratulations!")
```
